```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import tensorflow as tf
from tensorflow import keras
from keras import Sequential
from keras.layers import Dense
```

```python
df=pd.read_csv('/content/Social_Network_Ads.csv')
```

```python
df.head()
```

```python
# @title Correlation between Age and Estimated Salary

sns.scatterplot(x='Age', y='EstimatedSalary', hue='Gender', data=df)
```

```python
x=df.iloc[:,2:4]
```

```python
y=df.iloc[:,-1]
```

```python
# feature Scaling is a mandatory step
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test, y_train, y_test=train_test_split(x,y,test_size=0.2, random_state=101)
```

```python
scaler=StandardScaler()
X_train_sc=scaler.fit_transform(X_train)
X_test_sc=scaler.transform(X_test)
```

```python
dnn=Sequential()
dnn.add(Dense(units=32, activation='relu', input_dim=2))
dnn.add(Dense(units=16, activation='relu'))
dnn.add(Dense(units=1,activation='sigmoid'))
```

```python
from keras.callbacks import EarlyStopping
```

```python
callback=EarlyStopping(monitor='val_loss',min_delta=0.0001, patience=20, verbose=1, mode='auto')
```

```python
dnn.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])
dnn.fit(X_train_sc,y_train, validation_data=(X_test_sc,y_test), epochs=1000, callbacks=callback)
```
