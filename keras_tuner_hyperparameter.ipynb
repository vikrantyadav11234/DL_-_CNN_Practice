```python
from kerastuner import RandomSearch
```

```python

def build_model(hp):
  model=Sequential()
  model.add(Dense(32, activation='relu', input_dim=8))
  model.add(Dense(1, activation='sigmoid'))
  optimizer=hp.Choice('optimizer', values=['adam','sgd', 'rmsprop', 'adadelta'])
  model.compile(optimizer=optimizer, loss='binary_crossentropy', metrics=['accuracy'])
  return model
```

```python
from typing_extensions import Literal
# Best optimezer
tuner= kt.RandomSearch(build_model, objective="val_accuracy")
```

```python
tuner.search(X_train, y_train, epochs=5, validation_data=(X_test,y_test))
```

```python
tuner.get_best_hyperparameters()
```

```python
# Best optimizer
tuner.get_best_hyperparameters()[0].values
```

```python
model=tuner.get_best_models(num_models=1)[0]
```

```python
model.summary()
```

```python
model.fit(X_train, y_train, batch_size=32, epochs=100, initial_epoch=6,
           validation_data=( X_test, y_test), verbose=1)
```

# right numbers of neurons in the given layers- keras tuner hyperparameter

```python
def build_model(hp):
  model=Sequential()
  units=hp.Int('units',8,128, step=8)
  # 8-lower limit, 128-upper limit, step- iteration(every 8 points)
  model.add(Dense(units=units, activation='relu', input_dim=8))
  model.add(Dense(1, activation='sigmoid'))
  model.compile(optimizer='rmsprop', loss='binary_crossentropy', metrics=['accuracy'])
  return model
```

```python
tuner=kt.RandomSearch(build_model, objective='val_accuracy', max_trials=5,
                      directory='mydir', project_name='vikrant')
```

```python
tuner.search(X_train,y_train, epochs=5, validation_data=(X_test,y_test))
```

# No of hidden layers

```python
def build_model(hp):
  model=Sequential()
  model.add(Dense(64,activation='relu', input_dim=8))
  for i in range(hp.Int('num_hidden_layers', min_value=1, max_value=20)):
    model.add(Dense(64, activation='relu'))
  model.add(Dense(1, activation='sigmoid'))
  model.compile(optimizer='rmsprop', loss='binary_crossentropy', metrics=['accuracy'])
  return model

```

# Keras_tuner Hyperparameter tuning all at once

```python
def build_model(hp):
  model=Sequential()
  counter=0

  for i in range(hp.Int('num_of_layers', min_value=1, max_value=20)):
    if counter == 0:
      model.add(Dense(hp.Int('units'+str(i), min_value=8, max_value=256),
                      activation= hp.Choice('activation'+str(i),
                                            values=['relu', 'leaky_relu', 'tanh','softmax',
                                                    'sigmoid','linear','elu']),
                      input_dim=8))
      model.add(Dropout(hp.Choice('dropout'+str(i), values=[0.1,0.2,0.3,0.35,0.4,0.45,0.5])))
    else:
      model.add(Dense(hp.Int('units'+str(i), min_value=8, max_value=256),
                      activation= hp.Choice('activation'+str(i),
                                            values=['relu', 'leaky_relu', 'tanh','softmax',
                                                    'sigmoid','linear','elu'])))
      model.add(Dropout(hp.Choice('dropout'+str(i), values=[0.1,0.2,0.3,0.35,0.4,0.45,0.5])))
    counter+=1
  model.add(Dense(1, activation='sigmoid'))
  model.compile(optimizer=hp.Choice('optimizer',values=['rmsprop','adam','adagrad',
                                                        'adadelta','sgd','nadam']),
                                    loss='binary_crossentropy', metrics=['accuracy'] )
  return model
```

```python
tuner= kt.RandomSearch(build_model, objective='val_accuracy', max_trials=5,
                       directory='mydir', project_name='ALL1')
```

```python
tuner.search(X_train, y_train, epochs=5, validation_data=(X_test, y_test))
```

```python
tuner.get_best_hyperparameters()
```

```python
tuner.get_best_hyperparameters()[0].values
```

```python
model= tuner.get_best_models(num_models=1)[0]
```

```python
model.fit(X_train, y_train, epochs=100, initial_epoch=6,
          validation_data=(X_test, y_test))
```
