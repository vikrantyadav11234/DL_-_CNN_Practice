```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df=pd.read_csv('/content/placement.csv')
```

```python
df.head()
```

```python
sns.scatterplot(x=df['cgpa'],y=df['resume_score'],hue=df['placed'])
```

```python
x=df.iloc[:,0:2]
y=df.iloc[:,-1]
```

```python
x.head()
```

```python
y.head()
```

```python
# Mandotary steps- Feature Scaling
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
scaler=StandardScaler()
x_scaler=scaler.fit_transform(x)
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train,X_Test,y_train,y_test=train_test_split(x_scaler,y,test_size=0.2, random_state=1)
```

# Deep Neural Network/ Artificial Neural Network Model

```python
import tensorflow as tf
from tensorflow import keras
from keras import Sequential
from keras.layers import Dense
```

```python
dnn=Sequential()
dnn.add(Dense(units=6, activation= 'relu'))
dnn.add(Dense(units=6, activation= 'relu'))
dnn.add(Dense(units=1, activation='sigmoid'))
```

```python
dnn.compile(loss='binary_crossentropy',optimizer='adam')
dnn.fit(X_train,y_train, epochs=500, validation_data=(X_Test,y_test))
```
