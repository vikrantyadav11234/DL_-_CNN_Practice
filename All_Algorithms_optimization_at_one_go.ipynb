```python
import tensorflow as tf
from tensorflow import keras
from keras.datasets import mnist
from keras.models import Sequential
from keras.layers  import Dense, Dropout, Flatten
```

```python
(X_train, y_train),(X_test, y_test)=mnist.load_data()
print(X_train.shape, y_train.shape), print(X_test.shape,y_test.shape)
```

```python
# mnist data set is a gray dataset. so channel 1. we have to reshape it.
```

```python
X_train=X_train.reshape(X_train.shape[0],28,28,1)
X_test=X_test.reshape(X_test.shape[0],28,28,1)
input_shape=(28,28,1)
```

```python
print(X_train.shape, y_train.shape), print(X_test.shape,y_test.shape)
```

```python
X_train[1]
```

```python
## Mandatory step- normalization
```

```python
X_train=X_train/255.0
X_test=X_test/255.0
```

```python
y_train.dtype
```

```python
# its a multiclass classification bcoz output has 10 digits from 0 to 9
# so convert into categorical format
```

```python
y_train=keras.utils.to_categorical(y_train)
y_test=keras.utils.to_categorical(y_test)
```

```python
# build the model now
```

# mini batch gradient descent


```python
batch_size=64
num_classes=10
epochs= 10

def build_model(optimizer):
  model = Sequential()
  model.add(Flatten())
  model.add(Dense(256,activation='relu',input_shape=input_shape))
  model.add(Dropout(0.25))
  model.add(Dense(256, activation='relu'))
  model.add(Dropout(0.50))
  model.add(Dense(num_classes, activation='softmax'))
  model.compile(loss=keras.losses.categorical_crossentropy, optimizer=optimizer, metrics=['accuracy'])
  return model
```

# Train the model

```python
optimizers= ['SGD', 'Adagrad', 'Adadelta', 'RMSprop','Adam']

for i in optimizers:
  model= build_model(i)
  history= model.fit(X_train,y_train, validation_data=(X_test,y_test),batch_size=batch_size,
                     epochs=epochs, verbose=1)
```
